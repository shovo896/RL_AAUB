```python
import numpy as np 
import matplotlib.pyplot as plt 


def rc_filter(t,beta,T=1.0): 
    t=np.array(t,dtype=float)
    denom = 1 - (2*beta*t/T) ** 2 
    with np.errstate(divide='ignore',invalid='ignore'): 
        h=np.sinc(t/T) * np.cos(np.pi*beta*t/T)/denom 
    singular = np.isclose(denom,0.0)
    
    if beta > 0 : 
        h[singular]= (np.pi/4)*np.sinc(1/(2*beta))
    return h 


def rrc_filter(t,beta,T= 1.0): 
    t=np.asarray(t,dtype= float)
    h=np.zeros_like(t)
    
    for i, ti in enumerate(t): 
        if np.isclose(ti,0): 
            
            h[i] = (1/T) * (1 + beta * (4/np.pi - 1))
        elif beta > 0 and np.isclose(abs(ti),T/(4*beta)): 
            h[i] = (beta/T * np.sqrt(2)) * ((1 + 2/np.pi) * np.sin(np.pi/(4*beta)) + (1 - 2/np.pi) * np.cos(np.pi/(4*beta)))
            
        else : 
            num = np.sin(np.pi * ti / T * (1 - beta)) + 4 * beta * ti / T * np.cos(np.pi * ti / T * (1 + beta))
            den = np.pi * ti / T * (1 - (4 * beta * ti / T) ** 2)
            h[i] = num / den
    return h 

T = 1.0 
t = np.linspace(-6*T,6*T,2000)
beta = 0.35 



h_rc = rc_filter(t,beta,T)
h_rrc = rrc_filter(t,beta,T)
plt.plot(t/T,h_rc,label='RC')
plt.plot(t/T, h_rrc, label='RRC')
plt.legend() ; plt.grid(alpha = 0.3)
plt.xlabel("t/T")
plt.title(f"beta ={beta} ")
plt.show()
            
            
        
```
